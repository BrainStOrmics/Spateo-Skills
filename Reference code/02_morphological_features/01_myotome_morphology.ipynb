# 1. Load the left and right myotome datasets


## Load and validate data


```python
Myotome_left = st.read_h5ad("/DATA/User/gaomohan/Myotome_v2_left.h5ad")
Myotome_left
```

```python
Myotome_right = st.read_h5ad("/DATA/User/gaomohan/Myotome_v1_right.h5ad")
Myotome_right
```

# 1.1 Reconstruct the left-myotome point cloud


## Construct the point-cloud model


```python
Myotome_left_pc, plot_cmap = st.tdr.construct_pc(
    adata=Myotome_left.copy(),
    spatial_key="spatial",
    groupby="celltype",
    key_added="tissue",
    colormap="#00BFC4",
)
```

```python
st.pl.three_d_plot(
    model=Myotome_left_pc,
    key="tissue",
    model_style="points",
    show_axes=True,
    colormap=plot_cmap,
    jupyter="static",  # False
    window_size=(1024, 1024),
    cpo=cpo1,
    # filename = f"./figures/cs13.pdf"
)
```

# 1.2 Reconstruct the left-myotome surface


## Reconstruct the surface mesh


```python
Myotome_left_mesh, _, _ = st.tdr.construct_surface(
    pc=Myotome_left_pc,
    key_added="tissue",
    alpha=0.6,
    cs_method="marching_cube",
    cs_args={"mc_scale_factor": 1.5},
    smooth=5000,
    scale_factor=1.00,
)
```

```python
st.pl.three_d_plot(
    model=st.tdr.collect_models([Myotome_left_mesh, Myotome_left_pc]),
    key="tissue",
    model_style=["surface", "points"],
    show_axes=True,
    jupyter="static",
    window_size=(800, 800),
    cpo=cpo1,
)
```

```python
st.pl.three_d_plot(
    model=Myotome_left_mesh,
    key="tissue",
    model_style="surface",
    show_axes=True,
    jupyter="static",
    window_size=(800, 800),
    cpo=cpo1,
)
```

```python
st.pl.three_d_plot(
    model=Myotome_left_mesh,
    key="tissue",
    model_style="surface",
    show_axes=True,
    jupyter="static",
    window_size=(800, 800),
    cpo=cpo4,
)
```

# 2.1 Calculate right-myotome morphology


```python
um_right_pc_model, um_right_mesh_model = Myotome_right_pc.copy(), Myotome_right_mesh.copy()

um_right_pc_model.points = um_right_pc_model.points / 1000
um_right_mesh_model.points = um_right_mesh_model.points / 1000
```

## Calculate morphological features


```python
morph = st.tdr.model_morphology(model=um_right_mesh_model, pc=um_right_pc_model)
morph
```
