# Cell-Density Mapping on 3D Tissue Models

Estimate and visualize point-cloud kernel density for myotome, heart, and tail-bud models.

This curated notebook targets the current Dynamo-free Spateo API. Edit the path/configuration cells for a new system before execution.


```python
import warnings

warnings.filterwarnings("ignore")

import spateo as st
import pyvista as pv

pv.global_theme.transparent_background = True
```

# cs13 embryo


## Load and validate data


```python
Myotome_right = st.read_h5ad("/DATA/User/gaomohan/Myotome_v1_right.h5ad")
Myotome_right.uns["__type"] = "UMI"
Myotome_right
```

```python
Myotome_left = st.read_h5ad("/DATA/User/gaomohan/Myotome_v1_left.h5ad")
Myotome_left.uns["__type"] = "UMI"
Myotome_left
```

# cs13 Myotome point cloud model


## Construct the point-cloud model


```python
cpo = [
    (-1508.524414183453, -10974.135117384183, 25563.352608545407),
    (4720.1671, 112.69174999999996, 1200.0),
    (-0.6800775362520292, 0.7171120242306434, 0.15246274754575018),
]

Myotome_pc_1, plot_cmap = st.tdr.construct_pc(
    adata=Myotome_right.copy(),
    spatial_key="spatial",
    groupby="celltype",
    key_added="tissue",
    colormap="rainbow",
)

st.pl.three_d_plot(
    model=Myotome_pc_1,
    key="tissue",
    model_style="points",
    show_axes=True,
    colormap=plot_cmap,
    jupyter="static",  # False
    window_size=(800, 800),
    cpo=cpo,
)
```

```python
Myotome_pc_2, plot_cmap = st.tdr.construct_pc(
    adata=Myotome_left.copy(),
    spatial_key="spatial",
    groupby="celltype",
    key_added="tissue",
    colormap="rainbow",
)

st.pl.three_d_plot(
    model=Myotome_pc_2,
    key="tissue",
    model_style="points",
    show_axes=True,
    colormap=plot_cmap,
    jupyter="static",  # False
    window_size=(800, 800),
    cpo=cpo,
)
```

```python
cpo = [
    (-5006.789177572386, -25557.216094653202, -115.0783686672523),
    (4720.1671, 112.69174999999996, 1200.0),
    (-0.9334155956236307, 0.35598914126682585, -0.04480019136890938),
]
```

## Reconstruct the surface mesh


```python
Myotome_mesh_1, _, _ = st.tdr.construct_surface(
    pc=Myotome_pc_1,
    key_added="tissue",
    alpha=0.6,
    cs_method="marching_cube",
    cs_args={"mc_scale_factor": 1.6},
    smooth=6000,
    scale_factor=1.08,
)

st.pl.three_d_plot(
    model=st.tdr.collect_models([Myotome_mesh_1, Myotome_pc_1]),
    key="tissue",
    model_style=["surface", "points"],
    show_axes=True,
    jupyter="static",
    window_size=(400, 400),
    cpo=cpo,
    # filename = f"./figures/e9.5_heart_mesh.pdf",
)
```

```python
Myotome_mesh_2, _, _ = st.tdr.construct_surface(
    pc=Myotome_pc_2,
    key_added="tissue",
    alpha=0.6,
    cs_method="marching_cube",
    cs_args={"mc_scale_factor": 1.10},
    smooth=6000,
    scale_factor=1.08,
)

st.pl.three_d_plot(
    model=st.tdr.collect_models([Myotome_mesh_2, Myotome_pc_2]),
    key="tissue",
    model_style=["surface", "points"],
    show_axes=True,
    jupyter="static",
    window_size=(400, 400),
    cpo=cpo,
    # filename = f"./figures/e9.5_heart_mesh.pdf",
)
```

## Estimate cell density


```python
st.tdr.pc_KDE(pc=Myotome_pc_2, bandwidth=10, key_added="cells_kde", colormap="hot_r", inplace=True)
st.pl.three_d_plot(
    model=Myotome_pc_2,
    key="cells_kde",
    colormap="Blues_r",
    opacity=0.6,
    model_style="points",
    jupyter="static",
    window_size=(800, 800),
    cpo=cpo,
    # filename = f"./figures/e9.5_heart_cells_kde.pdf"
)
```

```python
st.tdr.pc_KDE(pc=Myotome_pc_2, bandwidth=20, key_added="cells_kde", colormap="hot_r", inplace=True)
st.pl.three_d_plot(
    model=Myotome_pc_2,
    key="cells_kde",
    colormap="hot_r",
    opacity=0.6,
    model_style="points",
    jupyter="static",
    window_size=(800, 800),
    cpo=cpo,
    # filename = f"./figures/e9.5_heart_cells_kde.pdf"
)
```
