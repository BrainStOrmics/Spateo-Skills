This curated notebook targets the current Dynamo-free Spateo API. Edit the path/configuration cells for a new system before execution.

## Load and validate data


```python
cs13 = st.read_h5ad("/DATA/User/gaomohan/DATA/CS13_Project/cs13/CS13.final.h5ad")
```

```python
cs13_down20 = st.read_h5ad(
    "/DATA/User/gaomohan/figures/figure3/b/downsmapling_aligned_cs13/p20_p40_p60_p80_corrected_20260825_v1/p20/CS13_p20_random_rigid_refined_SL175_SL199_z40um.h5ad"
)
cs13_down40 = st.read_h5ad(
    "/DATA/User/gaomohan/figures/figure3/b/downsmapling_aligned_cs13/p20_p40_p60_p80_corrected_20260825_v1/p40/CS13_p40_random_rigid_refined_SL175_SL199_z40um.h5ad"
)
cs13_down60 = st.read_h5ad(
    "/DATA/User/gaomohan/figures/figure3/b/downsmapling_aligned_cs13/p20_p40_p60_p80_corrected_20260825_v1/p60/CS13_p60_random_rigid_refined_SL175_SL199_z40um.h5ad"
)
cs13_down80 = st.read_h5ad(
    "/DATA/User/gaomohan/figures/figure3/b/downsmapling_aligned_cs13/p20_p40_p60_p80_corrected_20260825_v1/p80/CS13_p80_random_rigid_refined_SL175_SL199_z40um.h5ad"
)
```

```python
cs13, cs13_down20, cs13_down40, cs13_down60, cs13_down80
```

## Construct aligned downsampling point clouds


## Construct the point-cloud model


```python
cs13_pc, plot_cmap = st.tdr.construct_pc(
    adata=cs13.copy(),
    spatial_key="spatial",
    groupby="celltype",
    key_added="tissue",
    colormap="#00BFC4",
)

cs13_down20_pc, plot_cmap = st.tdr.construct_pc(
    adata=cs13_down20.copy(),
    spatial_key="aligned_spatial_3D",
    groupby="celltype",
    key_added="tissue",
    colormap="#00BFC4",
)

cs13_down40_pc, plot_cmap = st.tdr.construct_pc(
    adata=cs13_down40.copy(),
    spatial_key="aligned_spatial_3D",
    groupby="celltype",
    key_added="tissue",
    colormap="#00BFC4",
)

cs13_down60_pc, plot_cmap = st.tdr.construct_pc(
    adata=cs13_down60.copy(),
    spatial_key="aligned_spatial_3D",
    groupby="celltype",
    key_added="tissue",
    colormap="#00BFC4",
)

cs13_down80_pc, plot_cmap = st.tdr.construct_pc(
    adata=cs13_down80.copy(),
    spatial_key="aligned_spatial_3D",
    groupby="celltype",
    key_added="tissue",
    colormap="#00BFC4",
)
```

## Review the reference point cloud


```python
st.pl.three_d_plot(
    model=cs13_pc,
    key="tissue",
    model_style="points",
    model_size=1.2,
    opacity=0.8,
    ambient=0.6,
    colormap="#8F8D8D",
    background="white",  # White supports antialiasing; the exported PNG still has alpha=0.
    show_axes=False,
    show_legend=False,
    cpo=cpo,
    window_size=(1200, 1200),
    # Export to a PNG format that supports an alpha channel.
    off_screen=True,
    jupyter="static",
)
```

## Draw equal-size 75,000-cell samples


```python
n_sample = 75_000

idx_0 = st.tl.sample(
    arr=np.arange(cs13.n_obs),
    n=n_sample,
    method="random",
    seed=42,
)
cs13_75k = cs13[idx_0].copy()

idx_1 = st.tl.sample(
    arr=np.arange(cs13_down20.n_obs),
    n=n_sample,
    method="random",
    seed=42,
)
cs13_down20_75k = cs13_down20[idx_1].copy()

idx_2 = st.tl.sample(
    arr=np.arange(cs13_down40.n_obs),
    n=n_sample,
    method="random",
    seed=42,
)
cs13_down40_75k = cs13_down40[idx_2].copy()

idx_3 = st.tl.sample(
    arr=np.arange(cs13_down60.n_obs),
    n=n_sample,
    method="random",
    seed=42,
)
cs13_down60_75k = cs13_down60[idx_3].copy()

idx_4 = st.tl.sample(
    arr=np.arange(cs13_down80.n_obs),
    n=n_sample,
    method="random",
    seed=42,
)
cs13_down80_75k = cs13_down80[idx_4].copy()

cs13_down20_75k, cs13_down40_75k, cs13_down60_75k, cs13_down80_75k
```

```python
cs13_75k.write_h5ad("cs13_75k.h5ad")
cs13_down20_75k.write_h5ad("cs13_down20_75k.h5ad")
cs13_down40_75k.write_h5ad("cs13_down40_75k.h5ad")
cs13_down60_75k.write_h5ad("cs13_down60_75k.h5ad")
cs13_down80_75k.write_h5ad("cs13_down80_75k.h5ad")
```

## Construct equal-size point clouds


```python
cs13_75k_pc, plot_cmap = st.tdr.construct_pc(
    adata=cs13_75k.copy(),
    spatial_key="spatial",
    groupby="celltype",
    key_added="tissue",
    colormap="#00BFC4",
)

cs13_down20_75k_pc, plot_cmap = st.tdr.construct_pc(
    adata=cs13_down20_75k.copy(),
    spatial_key="aligned_spatial_3D",
    groupby="celltype",
    key_added="tissue",
    colormap="#00BFC4",
)

cs13_down40_75k_pc, plot_cmap = st.tdr.construct_pc(
    adata=cs13_down40_75k.copy(),
    spatial_key="aligned_spatial_3D",
    groupby="celltype",
    key_added="tissue",
    colormap="#00BFC4",
)

cs13_down60_75k_pc, plot_cmap = st.tdr.construct_pc(
    adata=cs13_down60_75k.copy(),
    spatial_key="aligned_spatial_3D",
    groupby="celltype",
    key_added="tissue",
    colormap="#00BFC4",
)

cs13_down80_75k_pc, plot_cmap = st.tdr.construct_pc(
    adata=cs13_down80_75k.copy(),
    spatial_key="aligned_spatial_3D",
    groupby="celltype",
    key_added="tissue",
    colormap="#00BFC4",
)
```

```python
st.pl.three_d_plot(
    model=cs13_down40_75k_pc,
    key="tissue",
    model_style="points",
    model_size=1.2,
    opacity=0.8,
    ambient=0.6,
    colormap="#8F8D8D",
    background="white",  # White supports antialiasing; the exported PNG still has alpha=0.
    show_axes=False,
    show_legend=False,
    cpo=cpo,
    window_size=(1200, 1200),
    # Export to a PNG format that supports an alpha channel.
    off_screen=True,
    jupyter="static",
)
```

## Reconstruct the surface mesh


```python
cs13_75k_mesh, _, _ = st.tdr.construct_surface(
    pc=cs13_75k_pc,
    key_added="tissue",
    alpha=0.6,
    cs_method="marching_cube",
    cs_args={"mc_scale_factor": 0.8},
    smooth=5500,
    scale_factor=1.04,
)
```

```python
st.pl.three_d_plot(
    model=cs13_75k_mesh,
    key="tissue",
    model_style="surface",
    show_axes=True,
    jupyter="static",
    window_size=(800, 800),
    cpo=cpo,
)
```

```python
cs13_down20_75k_mesh, _, _ = st.tdr.construct_surface(
    pc=cs13_down20_75k_pc,
    key_added="tissue",
    alpha=0.6,
    cs_method="marching_cube",
    cs_args={"mc_scale_factor": 0.8},
    smooth=5500,
    scale_factor=1.04,
)
```

```python
cs13_down40_75k_mesh, _, _ = st.tdr.construct_surface(
    pc=cs13_down40_75k_pc,
    key_added="tissue",
    alpha=0.6,
    cs_method="marching_cube",
    cs_args={"mc_scale_factor": 1.2},
    smooth=5500,
    scale_factor=1.04,
)
```

```python
cs13_down60_75k_mesh, _, _ = st.tdr.construct_surface(
    pc=cs13_down60_75k_pc,
    key_added="tissue",
    alpha=0.6,
    cs_method="marching_cube",
    cs_args={"mc_scale_factor": 0.8},
    smooth=5500,
    scale_factor=1.04,
)
```

```python
cs13_down80_75k_mesh, _, _ = st.tdr.construct_surface(
    pc=cs13_down80_75k_pc,
    key_added="tissue",
    alpha=0.6,
    cs_method="marching_cube",
    cs_args={"mc_scale_factor": 0.8},
    smooth=5500,
    scale_factor=1.04,
)
```

```python
score1 = st.tdr.pairwise_shape_similarity(
    model1_pcs=np.asarray(cs13_75k_mesh.points),
    model2_pcs=np.asarray(cs13_down20_75k_mesh.points),
)
print(score1)
```

```python
score2 = st.tdr.pairwise_shape_similarity(
    model1_pcs=np.asarray(cs13_75k_mesh.points),
    model2_pcs=np.asarray(cs13_down40_75k_mesh.points),
)
print(score2)
```

```python
score3 = st.tdr.pairwise_shape_similarity(
    model1_pcs=np.asarray(cs13_75k_mesh.points),
    model2_pcs=np.asarray(cs13_down60_75k_mesh.points),
)
print(score3)
```

```python
score4 = st.tdr.pairwise_shape_similarity(
    model1_pcs=np.asarray(cs13_75k_mesh.points),
    model2_pcs=np.asarray(cs13_down80_75k_mesh.points),
)
print(score4)
```
