# Bilateral Myotome 3D Reconstruction

Reconstruct left and right myotome point clouds and surfaces and compare their 3D geometry.

This curated notebook targets the current Dynamo-free Spateo API. Edit the path/configuration cells for a new system before execution.


```python
import pyvista as pv

pv.global_theme.transparent_background = True

import spateo as st
```

## Load and validate data


```python
Myotome_left = st.read_h5ad("/DATA/User/gaomohan/CS13_Project/cs13/myotome_v3_left.h5ad")
Myotome_left
```

```python
Myotome_right = st.read_h5ad("/DATA/User/gaomohan/Myotome_v1_right.h5ad")
Myotome_right
```

## Construct the point-cloud model


```python
Myotome_left_pc, plot_cmap = st.tdr.construct_pc(
    adata=Myotome_left.copy(),
    spatial_key="spatial",
    groupby="celltype",
    key_added="tissue",
    colormap="#00BFC4",
)
```

```python
st.pl.three_d_plot(
    model=Myotome_left_pc,
    key="tissue",
    model_style="points",
    show_axes=False,
    colormap=plot_cmap,
    jupyter="static",  # False
    window_size=(1024, 1024),
    cpo=cpo,
    # filename = f"./figures/cs13.pdf"
)
```

## Reconstruct the surface mesh


```python
Myotome_left_mesh, _, _ = st.tdr.construct_surface(
    pc=Myotome_left_pc,
    key_added="tissue",
    alpha=0.6,
    cs_method="marching_cube",
    cs_args={"mc_scale_factor": 1.12},
    smooth=5000,
    scale_factor=1.02,
)
```

```python
st.pl.three_d_plot(
    model=st.tdr.collect_models([Myotome_left_mesh, Myotome_left_pc]),
    key="tissue",
    model_style=["surface", "points"],
    show_axes=False,
    jupyter="static",
    window_size=(800, 800),
    cpo=cpo,
)
```

```python
st.pl.three_d_plot(
    model=Myotome_left_mesh,
    key="tissue",
    model_style="surface",
    show_axes=True,
    jupyter="static",
    window_size=(800, 800),
    cpo=cpo,
)
```

```python
st.pl.three_d_plot(
    model=Myotome_left_mesh,
    key="tissue",
    model_style="surface",
    show_axes=False,
    show_legend=False,
    jupyter="static",
    window_size=(800, 800),
    cpo=cpo,
)
```

```python
Myotome_right_pc, plot_cmap = st.tdr.construct_pc(
    adata=Myotome_right.copy(),
    spatial_key="spatial",
    groupby="celltype",
    key_added="tissue",
    colormap="#00BFC4",
)
```

```python
st.pl.three_d_plot(
    model=Myotome_right_pc,
    key="tissue",
    model_style="points",
    show_axes=True,
    colormap=plot_cmap,
    jupyter="static",  # False
    window_size=(1024, 1024),
    cpo=cpo,
    # filename = f"./figures/cs13.pdf"
)
```

```python
Myotome_right_mesh, _, _ = st.tdr.construct_surface(
    pc=Myotome_right_pc,
    key_added="tissue",
    alpha=0.6,
    cs_method="marching_cube",
    cs_args={"mc_scale_factor": 1.6},
    smooth=5000,
    scale_factor=1.02,
)
```

```python
st.pl.three_d_plot(
    model=st.tdr.collect_models([Myotome_right_mesh, Myotome_right_pc]),
    key="tissue",
    model_style=["surface", "points"],
    show_axes=True,
    jupyter="static",
    window_size=(800, 800),
    cpo=cpo,
)
```

```python
st.pl.three_d_plot(
    model=Myotome_right_mesh,
    key="tissue",
    model_style="surface",
    show_axes=True,
    jupyter="static",
    show_legend=False,
    window_size=(800, 800),
    cpo=cpo,
)
```

```python
st.pl.three_d_plot(
    model=Myotome_right_mesh,
    key="tissue",
    model_style="surface",
    show_axes=False,
    show_legend=False,
    jupyter="static",
    window_size=(800, 800),
    cpo=cpo,
)
```

```python
st.pl.three_d_plot(
    model=st.tdr.collect_models(
        [Myotome_right_mesh, Myotome_right_pc, Myotome_left_mesh, Myotome_left_pc]
    ),
    key="tissue",
    model_style=["surface", "points", "surface", "points"],
    show_axes=False,
    jupyter="static",
    window_size=(800, 800),
    show_legend=False,
    cpo=cpo,
)
```

```python
um_right_pc_model, um_right_mesh_model = Myotome_right_pc.copy(), Myotome_right_mesh.copy()

um_right_pc_model.points = um_right_pc_model.points / 1000
um_right_mesh_model.points = um_right_mesh_model.points / 1000
```

## Calculate morphological features


```python
morph = st.tdr.model_morphology(model=um_right_mesh_model, pc=um_right_pc_model)
morph
```
