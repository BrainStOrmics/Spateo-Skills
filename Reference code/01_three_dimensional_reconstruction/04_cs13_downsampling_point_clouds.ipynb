# CS13 Downsampling Point-Cloud Reconstruction

Create reproducible CS13 point-cloud models at multiple retained-cell fractions for reconstruction sensitivity analysis.

This curated notebook targets the current Dynamo-free Spateo API. Edit the path/configuration cells for a new system before execution.


```python
import numpy as np

# Enable transparent backgrounds.
import pyvista as pv

pv.global_theme.transparent_background = True
import spateo as st
```

```python
cpo = [
    (7235.672822135923, -9333.743725966886, 29537.530999873827),
    (5265.3827, 317.52565000000004, 1200.0),
    (-0.9910735612049205, -0.13108990476291582, 0.024261763123203748),
]
```

## Load and validate data


```python
cs13 = st.read_h5ad("/DATA/User/gaomohan/DATA/CS13_Project/cs13/CS13.final.h5ad")
```

```python
cs13
```

## Retain 80% of cells


```python
# Fix the random seed for reproducibility.
rng = np.random.default_rng(42)

n_sample = 1538427

if cs13.n_obs > n_sample:
    idx = rng.choice(cs13.n_obs, size=n_sample, replace=False)

    cs13_80_down = cs13[idx].copy()
else:
    cs13_80_down = cs13.copy()

print(cs13_80_down)
```

## Construct the point-cloud model


```python
cs13_80_pc, plot_cmap = st.tdr.construct_pc(
    adata=cs13_80_down.copy(),
    spatial_key="spatial",
    groupby="celltype",
    key_added="tissue",
    colormap="#00BFC4",
)
```

```python
st.pl.three_d_plot(
    model=cs13_80_pc,
    key="tissue",
    model_style="points",
    model_size=1.2,
    opacity=0.8,
    ambient=0.6,
    colormap="#8F8D8D",
    background="white",  # White supports antialiasing; the exported PNG still has alpha=0.
    show_axes=True,
    show_legend=False,
    cpo=cpo,
    window_size=(1200, 1200),
    # Export to a PNG format that supports an alpha channel.
    off_screen=True,
    jupyter="static",
)
```
