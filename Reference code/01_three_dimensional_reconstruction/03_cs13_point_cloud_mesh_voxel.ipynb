# CS13 Point-Cloud, Mesh, and Voxel Reconstruction

Build the principal CS13 model representations in the official point-cloud to surface to voxel order.

This curated notebook targets the current Dynamo-free Spateo API. Edit the path/configuration cells for a new system before execution.


```python
import spateo as st
```

## Load and validate data


```python
cs13 = st.read_h5ad("/DATA/User/gaomohan/cs13/CS13.final.h5ad")
cs13
```

## Construct the point-cloud model


```python
cs13_pc, plot_cmap = st.tdr.construct_pc(
    adata=cs13.copy(),
    spatial_key="spatial",
    groupby="celltype",
    key_added="tissue",
    colormap="#00BFC4",
)
```

```python
cpo = [
    (7235.672822135923, -9333.743725966886, 29537.530999873827),
    (5265.3827, 317.52565000000004, 1200.0),
    (-0.9910735612049205, -0.13108990476291582, 0.024261763123203748),
]
```

```python
st.pl.three_d_plot(
    model=cs13_pc,
    key="tissue",
    model_style="points",
    model_size=1.2,  # Use small points for the dense cloud.
    opacity=0.7,  # Slight transparency preserves the granular point-cloud appearance.
    ambient=0.6,  # Reduce lighting differences between rendered points.
    colormap="#AFAFAF",  # Use one neutral grey.
    background="white",
    show_axes=False,
    show_legend=False,
    jupyter="static",
    window_size=(1024, 1024),
    cpo=cpo,
)
```

```python
cs13_sub = st.read_h5ad("/DATA/User/gaomohan/cs13_sub_v1.h5ad")
```

```python
cs13_sub_pc, plot_cmap = st.tdr.construct_pc(
    adata=cs13_sub.copy(),
    spatial_key="spatial",
    groupby="celltype",
    key_added="tissue",
    colormap="#00BFC4",
)
```

```python
cpo = [
    (2309.8390732747557, -24500.600941468485, 18730.722256193923),
    (5366.7332, 307.3505499999999, 1200.0),
    (-0.9949720188641433, 0.0804198026782226, -0.059693693256546075),
]
```

```python
st.pl.three_d_plot(
    model=cs13_sub_pc,
    key="tissue",
    model_style="points",
    show_axes=True,
    colormap=plot_cmap,
    jupyter="static",  # False
    window_size=(1024, 1024),
    cpo=cpo,
    # filename = f"./figures/cs13.pdf"
)
```

## Reconstruct the surface mesh


```python
cs13_mesh, _, _ = st.tdr.construct_surface(
    pc=cs13_sub_pc,
    key_added="tissue",
    alpha=0.6,
    cs_method="marching_cube",
    cs_args={"mc_scale_factor": 0.98},
    smooth=5500,
    scale_factor=1.00,
)
```

```python
st.pl.three_d_plot(
    model=cs13_mesh,
    key="tissue",
    model_style="surface",
    show_axes=True,
    jupyter="static",
    window_size=(800, 800),
    cpo=cpo,
)
```

## Reconstruct the voxel model


```python
embryo_voxel, _ = st.tdr.voxelize_mesh(
    mesh=cs13_mesh,
    voxel_pc=None,
    key_added="tissue",
    label="embryo_voxel",
    color="gainsboro",
    smooth=500,
)
st.pl.three_d_plot(model=embryo_voxel, key="tissue", jupyter="static", cpo=cpo)
# st.tdr.save_model(model=embryo_voxel, filename="embryo_voxel_model.vtk")
```
