## Load and validate data


```python
Myotome_right = st.read_h5ad("/DATA/User/gaomohan/Myotome_v1_right.h5ad")
Myotome_right
```

## Construct the point-cloud model


```python
# Reconstruct point cloud model
Myotome_right_pc, _ = st.tdr.construct_pc(
    adata=Myotome_right.copy(), spatial_key="spatial", groupby="celltype"
)

st.tdr.add_model_labels(
    model=Myotome_right_pc,
    labels=["Point Cloud"] * Myotome_right_pc.n_points,
    key_added="backbone",
    where="point_data",
    inplace=True,
    alphamap=0.4,
    colormap="gainsboro",
)
```

## Reconstruct the surface mesh


```python
Myotome_right_mesh, _, _ = st.tdr.construct_surface(
    pc=Myotome_right_pc,
    alpha=0.3,
    cs_method="marching_cube",
    cs_args={"mc_scale_factor": 0.95},
    smooth=4000,
    scale_factor=1.0,
)

st.tdr.add_model_labels(
    model=Myotome_right_mesh,
    labels=["Mesh"] * Myotome_right_mesh.n_cells,
    key_added="backbone",
    where="cell_data",
    inplace=True,
    alphamap=0.3,
    colormap="gainsboro",
)
```

## Reconstruct the voxel model


```python
# Reconstruct voxel model
Myotome_right_voxel, _ = st.tdr.voxelize_mesh(
    mesh=Myotome_right_mesh,
    voxel_pc=None,
    key_added="backbone",
    label="Voxel",
    color="gainsboro",
    smooth=300,
)
```

```python
st.pl.three_d_multi_plot(
    model=st.tdr.collect_models(
        [st.tdr.collect_models([Myotome_right_pc, Myotome_right_mesh]), Myotome_right_voxel]
    ),
    key="backbone",
    model_style=[["points", "surface"], "surface"],
    jupyter="static",
    cpo=[cpo1],
    shape=(1, 2),
    ambient=[0.2, 0.1],
)
```

```python
import torch

from spateo.tdr.models.models_backbone.backbone_methods import NLPCA

def fixed_project(self, data):
    num_dim = data.shape[1]

    data_tensor = torch.tensor(
        data,
        dtype=torch.float32,
    )

    self.model.eval()

    with torch.no_grad():
        output = self.model(data_tensor)

        pts = output.detach().cpu().numpy()

        proj = self.model.intermediate_layer_model.detach().cpu().numpy()

    self.fit_points = pts

    all_data = np.concatenate(
        [pts, proj],
        axis=1,
    )

    all_sorted = all_data[all_data[:, num_dim].argsort()]

    return proj, all_sorted

NLPCA.project = fixed_project
```

## Reconstruct the anatomical backbone


```python
backbone_elpi, backbone_length_elpi, _ = st.tdr.construct_backbone(
    model=Myotome_right_voxel,
    rd_method="ElPiGraph",
    num_nodes=50,
    topology="curve",
    color="orangered",
)
```

```python
backbone, backbone_length, _ = st.tdr.construct_backbone(
    model=Myotome_right_voxel,
    rd_method="ElPiGraph",
    num_nodes=30,
    topology="curve",
    Lambda=0.01,
    Mu=0.2,
    color="orangered",
)
```

```python
from scipy.spatial import KDTree
```

```python
backbone_nodes = np.asarray(backbone.points)
backbone_nodes_kdtree = KDTree(np.asarray(backbone_nodes))
_, ii = backbone_nodes_kdtree.query(np.asarray(Myotome_right.obsm["spatial"]), k=1)
Myotome_right.obs["backbone"] = ii
```

## Associate gene expression with morphology


```python
st.tl.glm_degs(
    adata=Myotome_right,
    fullModelFormulaStr="~cr(backbone, df=3)",
    key_added="glm_degs",
    qval_threshold=0.05,
    llf_threshold=-1000,
)
print(Myotome_right.uns["glm_degs"]["glm_result"])
```

## Continuous A–P backbone-gene analysis

This section continues the original right Myotome backbone workflow with a continuous A→P projection. The 30-node ElPiGraph curve is ordered from the visually upper endpoint (A) to the lower endpoint (P) under the notebook's original `cpo1` camera. Every cell is projected to the complete polyline, and Spateo `glm_degs` tests the continuous position using `~cr(backbone_s, df=3)`.

Genes must be detected in at least 3% of cells before testing. Heatmap profiles are Gaussian-kernel smoothed over 200 A–P positions (`bandwidth=0.04`). In the representative-gene panels, point color encodes A–P position only and expression is encoded by y.

- Input: `/DATA/User/gaomohan/Myotome_v1_right.h5ad`
- Output: `/DATA/User/gaomohan/figures/figure3/d/Myotome_backbone_gene_analysis_20260821/right`


```python
import subprocess

analysis_root = Path(
    "/DATA/User/gaomohan/figures/figure3/d/Myotome_backbone_gene_analysis_20260821"
)
side_output = analysis_root / "right"
analysis_script = analysis_root / "code" / "myotome_backbone_gene_analysis.py"

cmd = [
    sys.executable,
    str(analysis_script),
    "--input-h5ad",
    "/DATA/User/gaomohan/Myotome_v1_right.h5ad",
    "--output-dir",
    str(side_output),
    "--side",
    "right",
    "--min-expressed-fraction",
    "0.03",
    "--reuse-glm-table",
]
# Reconstructs the backbone and reuses the validated GLM table when available.
subprocess.run(cmd, check=True)
```
