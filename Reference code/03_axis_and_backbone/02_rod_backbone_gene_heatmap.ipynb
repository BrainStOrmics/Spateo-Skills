## Reconstruct the voxel model


```python
# Reconstruct voxel model
stage1_ROD_voxel, _ = st.tdr.voxelize_mesh(
    mesh=stage1_ROD_mesh,
    voxel_pc=None,
    key_added="backbone",
    label="Voxel",
    color="gainsboro",
    smooth=300,
)
```

```python
import torch

from spateo.tdr.models.models_backbone.backbone_methods import NLPCA

def fixed_project(self, data):
    num_dim = data.shape[1]

    data_tensor = torch.tensor(
        data,
        dtype=torch.float32,
    )

    self.model.eval()

    with torch.no_grad():
        output = self.model(data_tensor)

        pts = output.detach().cpu().numpy()

        proj = self.model.intermediate_layer_model.detach().cpu().numpy()

    self.fit_points = pts

    all_data = np.concatenate(
        [pts, proj],
        axis=1,
    )

    all_sorted = all_data[all_data[:, num_dim].argsort()]

    return proj, all_sorted

NLPCA.project = fixed_project
```

```python
# PyVista line connectivity
lines = np.asarray(backbone.lines)

edges = []

i = 0
while i < len(lines):
    n = int(lines[i])

    ids = lines[i + 1 : i + 1 + n].astype(int)

    for a, b in zip(ids[:-1], ids[1:]):
        edges.append((a, b))

    i += n + 1

edges = np.asarray(edges)

# Count the number of edges incident on each node.
degree = np.zeros(backbone.n_points, dtype=int)

for a, b in edges:
    degree[a] += 1
    degree[b] += 1

endpoint_ids = np.where(degree == 1)[0]

print("Endpoint IDs:", endpoint_ids)

for idx in endpoint_ids:
    print(
        "endpoint:",
        idx,
        "node label:",
        backbone.point_data["nodes"][idx],
        "xyz:",
        backbone.points[idx],
    )
```

```python
endpoint_points = backbone.points[endpoint_ids]

p = pv.Plotter()

# Myotome
p.add_mesh(
    stage2_ROD_mesh,
    color="gainsboro",
    opacity=0.3,
)

# backbone
p.add_mesh(
    backbone,
    color="orangered",
    line_width=7,
)

# endpoint 1
p.add_points(
    endpoint_points[0:1],
    color="blue",
    model_size=20,
    render_points_as_spheres=True,
)

# endpoint 2
p.add_points(
    endpoint_points[1:2],
    color="green",
    model_size=20,
    render_points_as_spheres=True,
)

p.add_point_labels(
    endpoint_points,
    [
        f"Endpoint A\nnode {endpoint_ids[0]}",
        f"Endpoint B\nnode {endpoint_ids[1]}",
    ],
    font_size=18,
    model_size=15,
    always_visible=True,
)

p.show()
```

## 14 dpa ROD backbone-associated gene heatmaps

This section extends the 20-node backbone above as an independent, reproducible analysis that does not overwrite the source H5AD.

- **P** is identified automatically as node 9, the point with the greatest perpendicular distance from the A1-A2 endpoint chord.
- The combined axis is `0 -> 9 -> 19` (A1 to P to A2).
- The branch axes are `0 -> 9` (A1 to P) and `19 -> 9` (A2 to P).

### GLM parameter rationale

The original thresholds (`qval_threshold=0.01`, `llf_threshold=-1000`, `df=3`) retained only one gene. The revised analysis:

1. keeps genes detected in at least 1% of ROD cells to reduce uninformative tests;
2. uses `~cr(backbone_s, df=4)` for the combined U-shaped axis and `~cr(axis_progress, df=3)` for each half-axis;
3. sets `llf_threshold=None` because absolute log likelihood depends on expression abundance and sample size and is not an effect-size threshold;
4. saves complete GLM results and uses FDR <= 0.05 as the primary criterion, with a preregistered FDR <= 0.10 display threshold for the lower-power A1-P branch; and
5. plots mean size-factor-normalized, log1p expression per backbone node, followed by light axial smoothing, row-wise z scoring, and peak-position ordering.

The final analysis detected 27 genes on A1-P-A2, 10 genes on A1-P, and 27 genes on A2-P at FDR <= 0.05.

## Continuous AP-axis expression visualization

This section replaces the discrete backbone-node heatmap by Gaussian-kernel smoothing over each cell's continuous projection (`backbone_s`). The full axis is A1 → P → A2; branch-specific panels are A1 → P and A2 → P. In the selected-gene mini-plots, **point color encodes continuous AP position only**, expression is encoded by the y-coordinate, and the dark curve is the smoothed expression trend.

- Analysis input: `/DATA/User/gaomohan/figures/figure3/b&c&d/ROD_backbone_gene_heatmap_20260819`
- Continuous output: `/DATA/User/gaomohan/figures/figure3/b&c&d/ROD_backbone_gene_heatmap_continuous_20260819`
- Smoothing grid: 200 positions
- Full-axis bandwidth: 0.035; branch bandwidth: 0.050


```python
import subprocess

continuous_input_dir = Path(
    "/DATA/User/gaomohan/figures/figure3/b&c&d/ROD_backbone_gene_heatmap_20260819"
)
continuous_output_dir = Path(
    "/DATA/User/gaomohan/figures/figure3/b&c&d/ROD_backbone_gene_heatmap_continuous_20260819"
)
continuous_script = continuous_output_dir / "code" / "rod_continuous_backbone_heatmap.py"

cmd = [
    sys.executable,
    str(continuous_script),
    "--input-dir",
    str(continuous_input_dir),
    "--output-dir",
    str(continuous_output_dir),
    "--n-grid",
    "200",
    "--bandwidth-full",
    "0.035",
    "--bandwidth-branch",
    "0.050",
    "--n-key-full",
    "6",
    "--n-key-branch",
    "5",
]
subprocess.run(cmd, check=True)
```
